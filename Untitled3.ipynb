```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import os

# Unzip the uploaded file if the CSV does not exist yet
csv_filename = 'Sample - Superstore.csv'
zip_filename = 'Sample - Superstore.csv (1).zip'

if not os.path.exists(csv_filename):
    if os.path.exists(zip_filename):
        with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
            zip_ref.extractall('.')
        print(f"Extracted {zip_filename} successfully!")
    elif os.path.exists('Sample - Superstore.csv.zip'):
        with zipfile.ZipFile('Sample - Superstore.csv.zip', 'r') as zip_ref:
            zip_ref.extractall('.')
        print("Extracted Sample - Superstore.csv.zip successfully!")
    else:
        raise FileNotFoundError("Please make sure your zip or CSV file is uploaded.")

# File read karna (encoding Latin-1 use hoti hai superstore data ke liye)
df = pd.read_csv(csv_filename, encoding='windows-1252')

# Column names ke spaces clean karna
df.columns = df.columns.str.strip()

# Date format set karna
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M').astype(str)

print("Dataset successfully loaded! Total rows:", len(df))
print(df[['Order Date', 'Category', 'Sub-Category', 'Sales', 'Profit']].head())

# 1. Monthly Revenue Trend
monthly_sales = df.groupby('YearMonth')['Sales'].sum().reset_index()

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales['YearMonth'], monthly_sales['Sales'], marker='o', color='#1f77b4', linewidth=2)
plt.title('Monthly Sales Revenue Trend (Superstore)', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=11)
plt.ylabel('Total Sales ($)', fontsize=11)
plt.xticks(rotation=60)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('monthly_sales_trend.png', dpi=300)
plt.show()

# 2. Top Performing Sub-Categories by Sales
top_subcat = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=top_subcat, x='Sales', y='Sub-Category', palette='mako')
plt.title('Sales by Sub-Category', fontsize=14, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=11)
plt.ylabel('Sub-Category', fontsize=11)
plt.tight_layout()
plt.savefig('top_products.png', dpi=300)
plt.show()

# 3. Category & Regional Sales Breakdown
cat_region = df.groupby(['Region', 'Category'])['Sales'].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=cat_region, x='Region', y='Sales', hue='Category', palette='Set2')
plt.title('Regional Sales Breakdown by Category', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=11)
plt.ylabel('Total Sales ($)', fontsize=11)
plt.legend(title='Category')
plt.tight_layout()
plt.savefig('regional_sales.png', dpi=300)
plt.show()
```
